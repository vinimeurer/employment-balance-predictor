## 1 - Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# import seaborn as sns
```

## 2 - Configurações

```python
# ------------------------------
# Configurações do pandas
# ------------------------------

pd.set_option('display.max_columns', None)              # Mostrar todas as colunas
pd.set_option('display.max_rows', 100)                  # Mostrar mais linhas (opcional)
pd.set_option('display.float_format', '{:.2f}'.format)  # Melhor formatação de float
pd.set_option('display.expand_frame_repr', False)       # Evitar quebra de linha em colunas largas


# ------------------------------
# Configurações visuais
# ------------------------------

# sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# ------------------------------
# Caminhos
# ------------------------------

DATASET = "rmc.csv"
```

## 3 - Carregamento dos Dados

```python
df = pd.read_csv(DATASET)

df.head()
```

```python
# Shape
print(f"Linhas: {df.shape[0]}, Colunas: {df.shape[1]}")

# Tipos
df.info()

# Estatísticas
df.describe(include='all').T
```

```python
# Nulos
nulos = df.isnull().sum().sort_values(ascending=False)
nulos = nulos[nulos > 0]
print(nulos)

```
